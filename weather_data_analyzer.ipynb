##  Weather Data Analyzer

```python
class Weather:
    def __init__(self, region, temperature, humidity):
        self.region = region
        self.temperature = temperature
        self.humidity = humidity

# Function to classify weather condition based on temperature and humidity

    def classify_weather(self):
        if self.humidity > 80:
            return "Rainy"
        elif self.temperature > 30:
            return "Hot"
        elif 20 <= self.temperature <= 30:
            return "Warm"
        else:
            return "Cold"

 # Function to display region weather details

    def details_region(self):
        condition = self.classify_weather()  # Get weather condition
        print(f"Region: {self.region}")
        print(f"Temperature: {self.temperature}°C")
        print(f"Humidity: {self.humidity}%")
        print(f"Weather Condition: {condition}")
```

```python
# Function to calculate average temperature and humidity of all regions

def calculate_average(regions):
    total_temp = 0
    total_humidity = 0
    count = len(regions)

    # Loop through each region and sum temperature and humidity

    for region in regions:
        total_temp += region.temperature
        total_humidity += region.humidity

    # Calculate average values

    avg_temp = total_temp / count
    avg_humidity = total_humidity / count

    # Print the results

    print(f"\nAverage Temperature: {avg_temp:}")
    print(f"Average Humidity: {avg_humidity:}")
```

```python
regions = [
    Weather("city A", 28, 70),
    Weather("city B", 32, 65),
    Weather("city C", 26, 80),
    Weather("city D", 24, 85)
]
```

```python
region_input = input("Enter the region name to check weather: ")

found = False    # Check whether region exists in the list or not

# Loop through regions to find the matching region

for weather_data in regions:
    if weather_data.region.lower() == region_input.lower():
        print(f"\nWeather details for {region_input}:")
        weather_data.details_region()   # Display region details
        found = True
        break   # Exit loop once region is found

# If region not found, print message

if not found:
    print("Invalid region name")

else:
    # calculate average if the region is valid
    calculate_average(regions)
```
